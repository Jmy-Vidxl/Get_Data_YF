```python
import  os
import  time
import  yfinance                as      yf
import  numpy                   as      np
import  pandas                  as      pd
from    datetime                import  datetime, timedelta
```

```python
fil_dir = os.getcwd() + '/Input/List.xlsx'  # Filename direction
df_list = pd.read_excel( fil_dir, sheet_name = 'Lista' )
df_list
```

```python
n_colum = 1     # Usamos los datos asociados a ETFx 
df_name = list( ( df_list[ [df_list.columns[n_colum]] ].dropna().values ).reshape(-1) )

# Lista que usaremos para obtener los datos de YF
df_name
```

```python
# Descargamos los datos para cada Ticker y los guardamos en un diccionario
df_data = {}
for ticker in df_name:
    try:
        print( f"Downloading {ticker}..." )
        ticker_obj = yf.Ticker( ticker )
        df_data[ticker] = ticker_obj.history(period="max")
        time.sleep(3)   # Agregar un retraso para evitar el límite de solicitudes
    except Exception as e:
        print( f"Error downloading {ticker}: {e}" )
```

```python
# Guardar cada DataFrame en un archivo Excel con el nombre del Ticker
for ticker, df in df_data.items():
    if not df.empty:  # Verifica que el DataFrame no esté vacío
        fil_name = os.getcwd() + f"/Output/{df_list.columns[n_colum]}/{ticker}.xlsx"
        df.index = df.index.tz_localize(None)  # 🔹 Quitar la zona horaria de las fechas
        df.to_excel( fil_name )
        print( f"Guardado: {fil_name}" )
```
